```python
def plot_in_grid2(
        data,
        figsize=(30,20),
        col_width=4
    ):
    num_rows = len(data)

    fig = plt.figure(figsize=figsize)
    outer_gridspec = fig.add_gridspec(num_rows, 1)  # creates main vertical rows

    for row_index, row_plots in enumerate(data):
        # splits specific row into equal columns
        inner_gs = outer_gridspec[row_index].subgridspec(1, len(row_plots))

        for col_index, plot_data in enumerate(row_plots):
            ax = fig.add_subplot(inner_gs[0, col_index])

            if "title" in plot_data:
                ax.set_title(plot_data["title"])
            if "xlabel" in plot_data:
                ax.set_xlabel(plot_data["xlabel"])
            if "ylabel" in plot_data:
                ax.set_ylabel(plot_data["ylabel"])

            if "image" in plot_data:
                cmap = plot_data.get("cmap", "gray")
                img = ax.imshow(plot_data["image"], cmap=cmap)
                if plot_data.get("colorbar", True):
                    fig.colorbar(img, ax=ax, orientation="vertical")

            elif "curves" in plot_data:
                for curve in plot_data["curves"]:
                    x, y = curve[0], curve[1]
                    label = curve[2] if len(curve) > 2 else None
                    ax.plot(x, y, label=label)

                ax.legend()

    fig.tight_layout()
    return fig

# Usage
fig = plot_in_grid2(combined_subplots)
plt.show()
```
